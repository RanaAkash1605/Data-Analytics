# Array Attributes part-1

```python
# numpy attributes
import numpy as np

arr1 = np.arange(10,dtype=np.int32)
arr2 = np.arange(12,dtype=float).reshape(3,4)
arr3 = np.arange(8).reshape(2,2,2)

arr3
```

```python
# ndim : tells you the number of dimensions (axes) of a NumPy array.

arr3.ndim
```

```python
# shape : tells you how many elements are present in each dimension.

arr3.shape
```

```python
# size : tells you the total number of elements present in a NumPy array.

print(arr2.size)
arr2
```

```python
# itemsize : tells you how many bytes are used to store ONE element of a NumPy array.

arr3.itemsize
```

```python
# dtype means data type

print(arr1.dtype)
print(arr2.dtype)
print(arr3.dtype)
```

# Changing Data type

```python
# astype : used to convert the data type (dtype) of a NumPy array.

arr3.astype(np.int32)
```

# Array Operations

```python
a1 = np.arange(12).reshape(3,4)
a2 = np.arange(12,24).reshape(3,4)

a2
```

# 1. Scalar Opearations


```python

print(a1 + 5)
print(a1 - 6)
print(a1 * 3)
print(a1 / 3)
```

# 2. Vector Operations  

```python

""" In NumPy, vector operations mean performing mathematical operations
 directly on arrays/vectors without explicitly writing for loops. """

print(a1 + a2)
print(a2 - a1)
print(a1 * a2)
print(a1 / a2)

```

# Numpy Array Functions

```python
import numpy as np

ar1 = np.random.random((3,3))
ar1 = np.round(ar1*100) # np.round() is used to round decimal numbers to a specified number of decimal places.
ar1 # by default it is 0 but we can decide as well like (array, decimal places)
```

 # 1. max, min, sum, prod

```python
# There are important parameters that are used in these below functions axis and keepdims

# axis : Specifies the axis along which to perform the operation like (axis=0,1,None) # 0 means column-wised # 1 means row-wised

# keepdims : Retains reduced dimensions (True, False)


# max : In NumPy, the np.max() function is used to find the maximum (largest) value in an array.

max = np.max(ar1)
print(max)

# min : In NumPy, the np.min() function is used to find the minimum (smallest) value in an array.
min = np.min(ar1)
print(min)

# sum : The np.sum() function calculates the sum of all elements in an array or the sum along a specified axis.

sum = np.sum(ar1)
print(sum)

# prod : The np.prod() function calculates the product (multiplication) of all elements in an array or along a specified axis.
 
prod = np.prod(ar1)
print(prod)
```

# 3. mean, median, standard deviation and variance

```python
# There are important parameters that are used in these below functions axis and keepdims

# axis : Specifies the axis along which to perform the operation like (axis=0,1,None) # 0 means column-wised # 1 means row-wised

# keepdims : Retains reduced dimensions (True, False)


arr = np.array([20,40,60,80,100])

# mean
mean = np.mean(arr)   
print(mean)

# median
median = np.median(arr)
print(median)

# std
std = np.std(arr)
print(std)

# var
var = np.var(arr)
print(var)
```

# Trigonometry Functions

```python
# sin, cos, tan, arcsin, arccos, arctan

# arcsin : inverse sin

sin = np.sin(ar1)
print(sin)

cos = np.cos(ar1)
print(cos)

tan = np.tan(ar1)
print(tan)

inverse_sin = np.arcsin(ar1)
print(inverse_sin)

inverse_cos = np.arccos(ar1)
print(inverse_cos)

inverse_tan = np.arctan(ar1)
print(inverse_tan)
```
