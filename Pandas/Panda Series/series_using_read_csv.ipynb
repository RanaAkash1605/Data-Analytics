# Series Methods

```python
# head and tail

print(cars.head()) # gives data from top (by default 5)
print(cars.head(3)) # gives data from top
print(cars.tail()) # gives data from last (by default 5)
```

```python
# sample

print(cars.sample()) # random selected data
print(cars.sample(5)) # random selected data
```

```python
# value_counts

cars.value_counts() # return maximum values of data in dataset in descending order
```

```python
# sort_values 
# by default it is ascending
# by argument is used in this becoz it is dataframe not series
# it does not change the values of main dataset but if u want to change we can use inplace parameter

cars.sort_values(by="selling_price",ascending=False)
```

```python
# sort_index 
# this also use inplace parameter to change data permanently

cars.sort_index()
```

# Series Maths Methods

```python
# count

cars.count() # returns total number of values in series
```

```python
# sum and product

print(cars.sum()) # returns sum of all values
# print(cars.product())
```

```python
# mean median mode std var

# Mean of numeric columns
print(cars.mean(numeric_only=True))

# Median of numeric columns
print(cars.median(numeric_only=True))

# Mode of numeric columns
print(cars.mode(numeric_only=True))

# Std of numeric columns
print(cars.std(numeric_only=True))

# Var of numeric columns
print(cars.var(numeric_only=True))
```

```python
# min max

print("max: ",cars.max()) # max value
print("min: ",cars.min()) # min value
```

```python
# describedescribe() gives a quick statistical summary of a Series.

# For numeric data, it returns count, mean, standard deviation, quartiles, minimum and maximum.

# For categorical data, it returns count, unique, top and frequency.

# Missing values are excluded from the summary statistics.


cars.describe()
```

# Series Indexing

```python
# integer indexing

x = pd.Series([12,13,14,35,46,57,58,79,9])
x[1]
# x[-1] Negative indexing does not work in Series but it will work if indexes are string

cars.iloc[0]
```

```python
# Slicing

cars[5:16]
```

```python
cars[-5:]
```

```python
# Fancy indexing

print(cars.iloc[[1,3,4,5]]) # Selects by integer positions
print(cars.loc[[1,3,4,5]]) # Selects by index labels
```

# Editing indexing

```python
# using indexing

marks = {
    'maths':67,
    'english':78,
    'science':45,
    'hindi':90
}

marks_series = pd.Series(marks,name="Akash ke marks")
marks_series
```

```python
marks_series.iloc[1] = 100
marks_series.loc['english'] = 50
marks_series
```

```python
# What if an index does not exist

marks_series['sst'] = 90
marks_series
```

```python
# Slicing

marks_series[1:3] = [100,100]
marks_series
```

```python
# fancy indexing

marks_series.iloc[[0,2,3]] = [0,0,0]
marks_series
```
