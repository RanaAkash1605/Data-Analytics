# Series Attributes

-> size

-> dtype

-> name

-> is_unique

-> index

-> values

```python
import numpy as np
import pandas as pd

marks = {
    'maths':67,
    'english':78,
    'science':45,
    'hindi':90
}

marks_series = pd.Series(marks,name="Akash ke marks")
```

```python
# size

marks_series.size
```

```python
# dtype

marks_series.dtype
```

```python
# name

marks_series.name
```

```python
# is_unique

marks_series.is_unique
```

```python
# index

print(marks_series.index)
print(type(marks_series.index))
```

```python
# values

marks_series.values
```
