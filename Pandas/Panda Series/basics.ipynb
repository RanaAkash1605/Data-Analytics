# PANDA SERIES 

-> A one-dimensional labelled data structure used to store a single column of data.

# Importing Pandas

-> import pandas as pd

```python
import numpy as np
import pandas as pd
```

# Series from list

```python
# string

country = ['India','USA','Nepal','China']

pd.Series(country)
```

```python
# integer

runs = [13,43,12,45,76]

pd.Series(runs)
```

```python
# custom index

marks = [34,65,12,78]

subjects = ['maths','english','science','hindi']

pd.Series(marks,index=subjects)
```

```python
# setting a name
pd.Series(marks,index=subjects,name='Akash ke marks')
```

# Series from dict

```python
marks = {
    'maths':67,
    'english':78,
    'science':45,
    'hindi':90
}

marks_series = pd.Series(marks,name="Akash ke marks")
marks_series
```
