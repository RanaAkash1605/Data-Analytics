### Types of Unclean Data

-> Dirty Data (Data with Quality issues) : Dirty data, also known as low quality data. Low quality data has content issues.

- Duplicated data
- Missing data
- Corrupt data
- Inaccurate data

-> Messy Data (Data with tidiness issues) : Messy data, alos known as untidy data. Untidy data has structural issues. Tidy data has the following properties:

- Each variable forms a column
- Each observation forms a row
- Each observation unit forms a table
