## Disperion

-> Measure of spreads, variablity or scale of the data arround its central location.

-> Dispersion refers to the extent to which a distribution is stretched or squeezed.

-> Common measures of dispersion include range, variance, and standard deviation. 

```python

20 - 22.5  = -2.5 ^2 = 6.25
25 - 22.5  = 2.5 ^2 = 6.25
15 - 22.5  = -7.5 ^2 = 56.25
30 - 22.5  = 7.5 ^2 = 56.25
sigma^2 = (6.25 + 6.25 + 56.25 + 56.25) / 4 = 31.25
sigma = np.sqrt(sigma^2) = np.sqrt(31.25) = 5.5901699437494745

variance = sigma^2 = 31.25
standard_deviation = sigma = 5.5901699437494745
sample variance = s^2 = 31.25
sample standard_deviation = s = 5.5901699437494745
```
