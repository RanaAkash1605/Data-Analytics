```python
# Important Libraries

import numpy as np 
import pandas as pd
from scipy import stats # Used for Scientific and Mathematical computing like statistics. 
```

```python
# Mean

data = np.array([10,20,30,40,50])
Mean = np.mean(data)
print("Mean : ", Mean)
```

```python
# Median

data = np.array([10,20,30,40,50])
Median = np.median(data)
print("Median : ", Median)
```

```python
# Arithemtetic mean and median

np.random.seed(42) 
clean_data = np.random.normal(loc=6,scale=2.0,size=10)
print(clean_data)

corrupted_data = clean_data.copy() # Create a copy for potential corruption
corrupted_data[0] = 1e8 # Introduce coruption in the first element

# Get mean and median for both clean and corrupted data

clean_mean = np.mean(clean_data)
clean_median = np.median(clean_data)
corrupted_mean = np.mean(corrupted_data)
corrupted_median = np.median(corrupted_data)

print(f"Clean Mean :- {clean_mean}, Median :- {clean_median}")
print(f"Corrupted Mean :- {corrupted_mean}, Median :- {corrupted_median}")
```

```python
#Trimmed Mean(excluding or removing the top and bottom 10% of data)

trimmed_clean_mean = stats.trim_mean(clean_data, proportiontocut=0.1)
trimmed_corrupted_data = stats.trim_mean(corrupted_data, proportiontocut=0.1)

print(f"Trimmed Clean Mean :- {trimmed_clean_mean}")
print(f"Trimmed Corrupted Mean :- {trimmed_corrupted_data}")
```
