# Advance Matplotlib

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

# Colored Scatterplots

```python
iris = pd.read_csv(r'C:\data analytics\Matplotlib\iris.csv')
iris.sample(5)
```

```python
iris['Species'] = iris['Species'].replace({'Iris-setosa':0,'Iris-versicolor':1,'Iris-virginica':2})
iris.sample(5)
```

```python
plt.scatter(iris['SepalLengthCm'], iris['PetalLengthCm'],c=iris['Species'],cmap='jet',alpha=0.7)
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
plt.colorbar() # it show the color bar
```

# Plot Size

```python
plt.figure(figsize=(15,7)) # we can increase and decrease the size of plot or graph

plt.scatter(iris['SepalLengthCm'], iris['PetalLengthCm'],c=iris['Species'],cmap='jet',alpha=0.7)
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
plt.colorbar() 
```

# Annotations

```python
batters = pd.read_csv(r'C:\data analytics\Matplotlib\batter.csv')

# top 100 batsman

sample_df =batters.head(100).sample(25,random_state=5)
sample_df
```

```python
plt.figure(figsize=(18,10))
plt.scatter(sample_df['avg'], sample_df['strike_rate'],s=sample_df['runs'])

for i in range(sample_df.shape[0]):
    plt.text(sample_df['avg'].values[i], sample_df['strike_rate'].values[i], sample_df['batter'].values[i]) # text used to define or name the points.
```

# Horizontal and Vertical lines

```python
plt.figure(figsize=(18,10))
plt.scatter(sample_df['avg'], sample_df['strike_rate'],s=sample_df['runs'])

plt.axhline(130,color='red') # batsman which has strike rate more than 130
plt.axvline(30,color='red') # batsman which has strike rate more than 30

for i in range(sample_df.shape[0]):
    plt.text(sample_df['avg'].values[i], sample_df['strike_rate'].values[i], sample_df['batter'].values[i]) # text used to define or name the points.
```

# Sub Plots

```python
# a diff way to plot graphs
batters.head()
```

```python
# normal way

plt.figure(figsize=(15,6))
plt.scatter(batters['avg'],batters['strike_rate'])


plt.title('Something')
plt.xlabel('Avg')
plt.ylabel('Strike Rate')
plt.show()
```

```python
# new way or other way for subplots

# this method is help to plot multiple graphs at a time.

fig, ax = plt.subplots(figsize=(15,6))

ax.scatter(batters['avg'],batters['strike_rate'])
ax.set_title('Something')
ax.set_xlabel('Avg')
ax.set_ylabel('Strike Rate')
fig.show()
```

```python
# comparing two graphs

fig, ax = plt.subplots(nrows=2,ncols=1, sharex=True,figsize=(10,6)) # share x remove the x-axis label
ax[0].scatter(batters['avg'], batters['strike_rate'],color='red')
ax[1].scatter(batters['avg'], batters['runs'])

ax[0].set_title('Avg vs Strike Rate')
ax[0].set_ylabel('Strike Rate')

ax[1].set_title('Avg vs Runs')
ax[1].set_ylabel('Runs')
ax[1].set_xlabel('Avg')
fig.show()
```

```python
# we can add multiple plots 
fig, ax = plt.subplots(nrows=2,ncols=2,figsize=(10,10))

ax[0,0].scatter(batters['avg'], batters['strike_rate'],color='red')
ax[0,1].scatter(batters['avg'], batters['runs'])
ax[1,0].hist(batters['avg'])
ax[1,1].hist(batters['runs'])
```

```python
# we can take from above

fig = plt.figure()

ax1 = plt.subplot()
ax1.scatter(batters['avg'],batters['strike_rate'],color='red')
```

```python
# adding more plot 
fig = plt.figure()

ax1 = fig.add_subplot(2,2,1) # 2 - row , 2 - column , 1 - first graph 
ax1.scatter(batters['avg'],batters['strike_rate'],color='red')

ax2 = fig.add_subplot(2,2,2) 
ax2.hist(batters['runs'])

ax3 = fig.add_subplot(2,2,3)
ax3.hist(batters['avg'])
```
