```python
# linstyle

virat_performance = plt.plot(batsman['index'], batsman['V Kohli'],linestyle="dotted")
rohit_performance = plt.plot(batsman['index'], batsman['RG Sharma'],linestyle="dashdot")

plt.title('Rohit Sharma Vs Virat Kohli Career Comparison') # add title on top
plt.xlabel('Season') # print on x-axis
plt.ylabel('Runs Score') # print on y-axis
```

```python
# linewidth

virat_performance = plt.plot(batsman['index'], batsman['V Kohli'],linewidth=5)
rohit_performance = plt.plot(batsman['index'], batsman['RG Sharma'],linewidth=2)

plt.title('Rohit Sharma Vs Virat Kohli Career Comparison') # add title on top
plt.xlabel('Season') # print on x-axis
plt.ylabel('Runs Score') # print on y-axis
```

```python
# marker
# markersize

virat_performance = plt.plot(batsman['index'], batsman['V Kohli'],marker="+",markersize=10)
rohit_performance = plt.plot(batsman['index'], batsman['RG Sharma'],marker=".",markersize=5)

plt.title('Rohit Sharma Vs Virat Kohli Career Comparison') # add title on top
plt.xlabel('Season') # print on x-axis
plt.ylabel('Runs Score') # print on y-axis
```

```python
# legend -> location

virat_performance = plt.plot(batsman['index'], batsman['V Kohli'],label="Virat")
rohit_performance = plt.plot(batsman['index'], batsman['RG Sharma'],label="Rohit")

plt.title('Rohit Sharma Vs Virat Kohli Career Comparison') # add title on top
plt.xlabel('Season') # print on x-axis
plt.ylabel('Runs Score') # print on y-axis

plt.legend(loc='best') # location parameter we can place the label anywhere in the graph. 
```

# Scatter Plots

-> Bivariate Analysis

-> numerical vs numerical

-> Use case - Finding correlation

```python
# plt scatter function

x = np.linspace(-10,10,50)

y = 10*x + 3 + np.random.randint(0,300,50)

plt.scatter(x,y)
```

```python
# plt.scatter on pandas dataframe

batter = pd.read_csv(r'C:\data analytics\Matplotlib\batter.csv')
btr = batter.head(50)
```

```python
# scatter(x-axis,y-axis) 
# marker

plt.scatter(btr['avg'],btr['strike_rate'],color="red",marker="+")
plt.title('Avg and SR analysis of Top 50 players')
plt.xlabel('Average')
plt.ylabel('Strike Rate')
```

# plt.plot vs scatter plot

-> scatter : slower technique(good for small dataset)

-> plt : faster technique (if there is huge dataset we can use plt.plot with scatter)

```python
# conversion (plt plot to scatter plot)

# -> if we add additional parameter in plt plot then it will converted into scatter plot

plt.plot(batsman['V Kohli'],batsman['RG Sharma'],'o') # o it is the 3rd parameter which converts plt plot to scatter plot
```

# BAR CHART

-> Bivariate Analysis

-> Numerical vs Categorical

-> Use case - Aggregate analysis of groups

```python
# simple bar chart

children =[40,20,50,10,30]
colors = ['red','blue','yellow','pink','black']

plt.bar(colors,children)
# color
plt.bar(colors,children,color='black')
```

```python
# horizontal bar chart

plt.barh(colors,children,color='black')
```

```python
df = pd.read_csv(r'C:\data analytics\Matplotlib\batsman_season_record.csv')
df
```

```python
# multiple bar charts

plt.bar(np.arange(df.shape[0]) - 0.2, df['2015'],width=0.2)
plt.bar(np.arange(df.shape[0]), df['2016'],width=0.2)
plt.bar(np.arange(df.shape[0]) + 0.2, df['2017'],width=0.2)

plt.xticks(np.arange(df.shape[0]), df['batsman'])
plt.show() # removes the information
```

```python
# a problem

children =[40,20,50,10,30]
colors = ['red red red red red','blue blue blue blue blueblue blue','yellow yellowyellowyellow','pink pink','black black black black']

plt.bar(colors,children, color='black')
plt.xticks(rotation='vertical')
plt.show()
```

```python
# Stacked bar chart

plt.bar(df['batsman'],df['2015'], label='2015')
plt.bar(df['batsman'],df['2016'],bottom=df['2015'], label='2016')
plt.bar(df['batsman'],df['2017'],bottom=df['2015'] + df['2016'], label='2017')

plt.legend()
plt.show()
```

# Pie Chart

-> Univariate and Bivariate Analysis

-> Categorical vs numerical

-> Use case - To find contribution on a standard cosine

```python
# simple data

data = [23,45,100,20,49]
subjects = ['english','maths','science','hindi','sst']

plt.pie(data, labels=subjects)
plt.show()
```

```python
# dataset

df = pd.read_csv(r'C:\data analytics\Matplotlib\gayle-175.csv')
df
```

```python
plt.pie(df['batsman_runs'], labels=df['batsman'], autopct='%0.1f%%')
plt.show()
```

```python
# percentage and colors

# percentage : autopct='%0.1f%%'

plt.pie(df['batsman_runs'], labels=df['batsman'], autopct='%0.1f%%',colors=['blue','green','yellow','pink','cyan','brown'])
plt.show()
```

```python
# explode shadow

plt.pie(df['batsman_runs'], labels=df['batsman'], autopct='%0.1f%%',explode=[0.3,0,0,0,0,0.1],shadow=True)
plt.show()
```
