# Reshaping, Transporse, Ravel

-> Reshaping is the process of changing the dimensions of an array without changing the number or order of its elements.

-> Transpose is the operation of interchanging the axes of an array. For a 2D array, it swaps rows and columns.

-> Ravel converts a multidimensional array into a one-dimensional array.

```python
a = np.array([1, 2, 3, 4, 5, 6])

b = np.reshape(a, (2, 3)) # reshape

print(b)

trans = np.transpose(b)
#a.T   # transporse  we can write like this also - array_name.T
print("transporse:\n ", trans)  


ravel = np.ravel(a) # multi-dimensional to 1D
print("Ravel: \n", ravel)
```

# Stacking and Splitting in NumPy

-> Stacking and splitting are NumPy operations used to combine multiple arrays into a single array or divide an array into multiple smaller arrays.

Stacking: Combines two or more arrays.

Splitting: Divides an array into multiple smaller arrays.

# Stacking

```python
# np.stack() : np.stack() joins multiple arrays along a new axis.
# stack = np.stack(a, axis=0)

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

c = np.stack((a, b)) # 1D Array

print(c)
print(c.shape)

# np.vstack() : np.vstack() stacks arrays vertically, meaning it combines them row-wise.
# np.vstack(tup)

d = np.vstack((a, b)) # 1D array

print(d)
print(d.shape) 

e = np.array([[1, 2],   # 2D array
              [3, 4]])

f = np.array([[5, 6],
              [7, 8]])

print(np.vstack((e, f)))


# np.hstack() : np.hstack() stacks arrays horizontally.
# np.hstack(tup)

g = np.array([1, 2, 3])  # 1D array
h = np.array([4, 5, 6])

i = np.hstack((g, h))

print(i)


j = np.array([[1, 2],
              [3, 4]])

k = np.array([[5, 6],
              [7, 8]])

print(np.hstack((j, k)))
```

# Splitting

```python
# np.split() : np.split() divides an array into multiple sections.
# np.split(ary, indices_or_sections, axis=0)

a = np.array([10, 20, 30, 40, 50, 60])

b = np.split(a, 3)

print(b)
```

```python
# Split using indices

a = np.array([10, 20, 30, 40, 50, 60])

b = np.split(a, [2, 4]) # 2 and 4 are split positions

print(b)
```

```python
# Split a 2D array

a = np.array([
    [1, 2],
    [3, 4],
    [5, 6],
    [7, 8]
])

b = np.split(a, 2, axis=0)

for x in b:
    print(x)
```

```python
# np.array_split(): divides an array into a specified number of sections, even if the sections cannot all be equal in size.
# np.array_split(ary, indices_or_sections, axis=0)

a = np.array([10, 20, 30, 40, 50])

b = np.array_split(a, 3)

print(b)
```

```python
# np.vsplit() splits an array vertically, along the row axis.
# np.vsplit(ary, indices_or_sections)

a = np.array([
    [1, 2],
    [3, 4],
    [5, 6],
    [7, 8]
])

b = np.vsplit(a, 2)

print(b)
```

```python
# np.hsplit() splits an array horizontally, along the column axis for 2D arrays.
# np.hsplit(ary, indices_or_sections)

a = np.array([
    [1, 2, 3, 4],
    [5, 6, 7, 8]
])

b = np.hsplit(a, 2)

print(b)
```
