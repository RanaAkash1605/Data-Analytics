# INDEXING AND SLCING

```python
import numpy as np 

a1 = np.arange(10)
a2 = np.arange(12).reshape(3,4)
a3 = np.arange(8).reshape(2,2,2)
```

# Indexing

```python
print(a1) # 1D array
print(a1[-1])
```

```python
# 2D array : In a 2D array, you use two indices: one for the row and one for the column
# array[row_index, column_index]

print(a2) 
print("indexing: ", a2[2][:])
print("indexing: ", a2[2][3])
print("indexing: ", a2[1,3]) # this method is only for numpy array
```

```python
# 3D array 

print(a3)
print("Indexing: ", a3[1,0,1])
print("Indexing: ", a3[0,1,0])
```

# Slicing

```python
# (start, stop, steps) # 1D array
print(a1)
print(a1[2:8])
print(a1[2:8:2])
print(a1[0:])
print(a1[:-1])
```

```python
# 2D array # (row, column)

print(a2)
print("Slicing: ",a2[:,2])
print("Slicing: ",a2[0,:])
print("Slicing: ",a2[1:,1:3])
print("Slicing: ",a2[1:,1:])
print("Slicing: ",a2[::2,::3])
```

```python
# 3D array

a3 = np.arange(27).reshape(3,3,3) # (2d array, row, column)

print(a3) 
print("Slicing: \n\n",a3[1])
print("Slicing: \n\n",a3[::2])
print("Slicing: \n\n",a3[1,:,1])
print("Slicing: \n\n",a3[::2,2,2])
print("Slicing: \n\n",a3[::2,-1,-1])
```
