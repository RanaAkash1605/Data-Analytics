# Numpy array v/s Python list 

# speed

```python
a = [i for i in range(10000000)]
b = [i for i in range(10000000, 20000000)]

c = []

import time

start = time.time()

for i in range(len(a)):
    c.append(a[i] + b[i])

end = time.time()

print("Execution time:", end - start, "seconds")
```

```python
#numpy : numpy is much faster than python list

import numpy as np
import time

a = np.arange(10000000)
b = np.arange(10000000,20000000)

start = time.time()

c = a + b

end = time.time()

print("Execution time:", end - start, "seconds")
```

# memory

```python
# list

a = [i for i in range(10000000)]

import sys

sys.getsizeof(a)
```

```python
# numpy : we can reduce the size of memory by using dtype= np.int32,16,8,3 like that
a = np.arange(10000000, dtype=np.int64) 
sys.getsizeof(a)
```

# Advance Indexing

```python
# Normal Indexing and Slicing

import numpy as np

a = np.arange(12).reshape(4,3)
print(a)
print("indexing: ", a[2])
print("Slicing: ", a[5:])
```

# fancy indexing

```python
a = np.arange(24).reshape(6,4)
print(a)
print("row-wise:\n ",a[[0,2,3]]) # if i need 0th, 2nd and 3rd row in the array
print("column-wise:\n ",a[:,[0,2,3]]) # if i need columns
```

# boolean indexing

```python
import numpy as np

# randint() is a function used to generate random integers. In NumPy, it belongs to the random module.
# np.random.randint(low, high=None, size=None)

a = np.random.randint(1, 100, 24).reshape(6,4) # how many number i want in 3rd parameter : 24
a
```

```python
# Find all numbers greater than 50

print(a > 50) # it gives boolean array
print(a[a > 50]) # it gives numbers more than 50
```

```python
# Find out even numbers

print(a[a % 2 == 0])
```

```python
# Find greater than 50 and are even

print(a[(a > 50) & (a % 2 == 0)]) # when work with boolean we have to use bitwise operators.
# like instead of and, we use &
```

```python
# Find all numbers not divisible by 7

print(a[a % 7 != 0])
# print(a[~(a % 7 == 0)]) we can write this as well
```
